# 2. Homework

```python
homework = pd.read_csv(dataset + "homework.csv")
```

```python
homework.head()
```

```python
homework.info()
```

```python
all_columns = homework.columns.tolist()
for column in all_columns:
    print(homework[column].value_counts())
```

```python
homework['Due_Date'] = pd.to_datetime(homework['Due_Date'], format='mixed')
homework['Due_Date'].dtypes
```

```python
homework = homework.drop(columns=['Assignment_Name'])
homework.head()
```

```python
homework['Status'].value_counts()
```

```python
homework['Status'] = homework['Status'].str.strip()

map_status = {
    'Done': 0,
    'not done': 1,
    '❌': 1,
    '✅': 0,
    '✔': 0,
    'pending': 2
}
homework['Status'] = homework['Status'].map(map_status)
homework['Status'].value_counts()
```

```python
homework['Status'].astype('int64')
```

```python
homework.head()
```

```python
homework['Grade_Feedback'].value_counts()
```

```python
homework['Guardian_Signature'] = homework['Guardian_Signature'].replace({'Yes': 1, 'No': 0, ' ' : pd.NA})
homework['Guardian_Signature'].value_counts(dropna = False)
```

```python
homework.to_csv("../datasets/homework_clean.csv", index=False)
```

# 3. Performance

```python
performance = pd.read_csv(dataset + "performance.csv")

performance.head()
```

```python
performance.info()
```

```python
performance['Subject'].value_counts()
```

```python
score_copy = performance['Exam_Score'].copy().astype(str).str.replace(r'\d', '9', regex=True)
score_copy.value_counts()
```

```python
invalid_scores = performance['Exam_Score'][(performance['Exam_Score']< 1) | (performance['Exam_Score']> 100)]
print(invalid_scores)
print(len(invalid_scores))
print("Minimum:", invalid_scores.min())
print("Maximum:", invalid_scores.max())
```

```python
hw_completion_copy = performance['Homework_Completion_%'].copy().astype(str).str.replace(r'\d', '9', regex=True)
print(hw_completion_copy.value_counts())
```

```python
performance['Homework_Completion_%'] = (performance['Homework_Completion_%'].str.replace('%', '', regex=False).astype(float))

performance.loc[performance['Homework_Completion_%'] < 0,'Homework_Completion_%'] = float('nan')

print("Minimum:", performance['Homework_Completion_%'].min())
print("Maximum:", performance['Homework_Completion_%'].max())
```

```python
performance['Homework_Completion_%'].dtypes
```

```python
hw_completion_cleaned = performance['Homework_Completion_%'].copy().astype(str).str.replace(r'\d', '9', regex=True)
print(hw_completion_cleaned.value_counts())
```

```python
performance.to_csv("../datasets/performance_clean.csv", index=False)
```

# 4. Students

```python
students = pd.read_csv(dataset + "students.csv")
students.head()
```

```python
birth_copy = pd.to_datetime(students['Date_of_Birth'], format='mixed')
birth_copy.dtypes
```

```python
students['Date_of_Birth'] = pd.to_datetime(students['Date_of_Birth'], format='mixed')
students = students.drop(columns=['Emergency_Contact'])
```

```python
grade_level_copy = students['Grade_Level'].copy().astype(str).str.replace(r'\d', '9', regex=True)
grade_level_copy.value_counts()
```

```python
students['Grade_Level'] = (students['Grade_Level'].str.replace('Grade ', '', regex=False).astype('int64'))
students['Grade_Level'].dtypes
```

```python
ID_copy = students['Student_ID'].copy().astype(str).str.replace(r'\d', '9', regex=True)
ID_copy.value_counts()
```

```python
students.info()
```

```python
students.to_csv("../datasets/students_clean.csv", index=False)
```

# 2. Homework

```python
homework = pd.read_csv(dataset + "homework.csv", na_values=[' '])
```

```python
homework.head()
```

```python
homework.info()
```

```python
homework['Due_Date'] = pd.to_datetime(homework['Due_Date'], format='mixed')
homework['Due_Date'].dtypes
```

```python
homework['Subject'].unique()
```

```python
homework = homework.drop(columns=['Assignment_Name'])
homework.head()
```

```python
homework['Status'].value_counts()
```

```python
homework['Status'] = homework['Status'].str.strip()

map_status = {
    'Done': 0,
    'not done': 1,
    '❌': 1,
    '✅': 0,
    '✔': 0,
    'pending': 1
}
homework['Status'] = homework['Status'].map(map_status)
homework['Status'].value_counts()
```

```python
homework['Status'].astype('int64')
```

```python
homework.head()
```

```python
homework['Grade_Feedback'].value_counts()
```

```python
homework['Guardian_Signature'] = homework['Guardian_Signature'].replace({'Yes': 1, 'No': 0})
homework['Guardian_Signature'].value_counts()
```

```python
homework = homework.drop(columns='Guardian_Signature')
homework.head()
```

```python
homework.to_csv("../datasets/homework.csv", index=False)
```
